## Reference solution.R

If you are stuck or want to compare approaches, the exact frozen reference solution is below. Run **Setup** again before using it if you want a clean workspace.

````r
#!/usr/bin/env Rscript
# This script computes gene network statistics from expression correlations

# Read expression count data
expr <- read.table('inputs/counts.tsv', header = TRUE, sep = '\t', row.names = 1)

# Log2-transform the data with log2(count + 1) as specified
log_expr <- log2(expr + 1)

# Compute gene-gene Pearson correlation matrix
cor_matrix <- cor(t(log_expr), method = 'pearson')

# Define network edges as gene pairs with absolute correlation above 0.7
threshold <- 0.7
adjacency <- abs(cor_matrix) > threshold
diag(adjacency) <- FALSE  # Remove self-loops

# Compute degree (number of connections per gene)
degree <- rowSums(adjacency)

# Compute mean absolute correlation for each gene
mean_abs_cor <- rowMeans(abs(cor_matrix) - diag(nrow(cor_matrix)))

# Create output data frame with three columns: gene, degree, and mean_correlation
degree_df <- data.frame(
  gene = names(degree),
  degree = degree,
  mean_correlation = mean_abs_cor
)

# Save results to outputs/ppi_stats.csv
write.csv(degree_df, 'outputs/ppi_stats.csv', row.names = FALSE)
````
