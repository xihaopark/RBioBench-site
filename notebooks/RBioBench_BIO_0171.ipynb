## Reference solution.R

If you are stuck or want to compare approaches, the exact frozen reference solution is below. Run **Setup** again before using it if you want a clean workspace.

````r
#!/usr/bin/env Rscript
# CPM normalization adjusts for sequencing depth differences between samples

# Read raw count data from input file
counts <- read.table('inputs/counts.tsv', header = TRUE, sep = '\t', row.names = 1)

# Compute CPM normalization: (count / total_count_per_sample) * 1e6
# This normalizes each count by the library size (total counts per sample)
cpm <- sweep(counts, 2, colSums(counts), FUN = '/') * 1e6

# Save CPM normalized counts to output file
write.csv(cpm, 'outputs/cpm.csv', row.names = TRUE)
````
