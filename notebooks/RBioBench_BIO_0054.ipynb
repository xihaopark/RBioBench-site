## Reference solution.R

If you are stuck or want to compare approaches, the exact frozen reference solution is below. Run **Setup** again before using it if you want a clean workspace.

````r
counts_path <- file.path("inputs", "counts.tsv")
if (!file.exists(counts_path)) stop("counts.tsv missing")
expr_df <- read.delim(counts_path, check.names = FALSE, row.names = 1)
expr_mat <- as.matrix(expr_df)
median_vals <- apply(expr_mat, 1, median)
mad_vals <- apply(expr_mat, 1, mad)
iqr_vals <- apply(expr_mat, 1, function(x) diff(quantile(x, probs = c(0.25, 0.75), na.rm = TRUE)))
summary_df <- data.frame(
  gene = rownames(expr_mat),
  median_expression = median_vals,
  mad_expression = mad_vals,
  iqr_expression = iqr_vals,
  stringsAsFactors = FALSE
)
outputs_dir <- "outputs"
dir.create(outputs_dir, showWarnings = FALSE)
write.csv(summary_df, file.path(outputs_dir, "row_medians.csv"), row.names = FALSE)
````
